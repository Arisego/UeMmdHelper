```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# CSV file path
# ============================================================
# Modify it to your exported csv
csv_path = r"F:\xxx.csv"


df = pd.read_csv(csv_path)

x = df["index"]


# ============================================================
# Feature group
# ============================================================

feature_groups = {

    "Energy": [
        "rms_energy",
        "bass_energy",
        "low_mid_energy",
        "mid_energy",
        "high_mid_energy",
        "treble_energy",
    ],

    "Normalized": [
        "rms_norm",
        "bass_norm",
        "low_mid_norm",
        "mid_norm",
        "high_mid_norm",
        "treble_norm",
    ],

    "Ratio": [
        "bass_ratio",
        "low_mid_ratio",
        "mid_ratio",
        "high_mid_ratio",
        "treble_ratio",
    ],

    "Spectral": [
        "spectral_centroid",
        "spectral_centroid_norm",
        "spectral_bandwidth",
        "spectral_bandwidth_norm",
        "spectral_rolloff",
        "spectral_rolloff_norm",
    ],

    "Onset / Flux": [
        "onset",
        "onset_norm",
        "spectral_flux",
        "spectral_flux_norm",
    ],

    "Rhythm": [
        "beat",
        "bpm",
    ],
}


# ============================================================
# Draw every group
# ============================================================

for group_name, columns in feature_groups.items():

    # Keep existing only
    columns = [
        col for col in columns
        if col in df.columns
    ]

    if not columns:
        continue

    fig, axes = plt.subplots(
        len(columns),
        1,
        figsize=(16, 3.5 * len(columns)),
        squeeze=False
    )

    axes = axes.flatten()

    for ax, column in zip(axes, columns):

        y = pd.to_numeric(
            df[column],
            errors="coerce"
        )

        ax.plot(
            x,
            y,
            marker=".",
            markersize=3,
            linewidth=1
        )

        ax.set_title(column)
        ax.set_xlabel("Index")
        ax.set_ylabel("Value")

        ax.grid(True, alpha=0.3)

    fig.suptitle(
        group_name,
        fontsize=18
    )

    plt.tight_layout()
    plt.subplots_adjust(top=0.93)

    plt.show()
```
